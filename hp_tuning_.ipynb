```python
import os
import pickle

import pandas as pd
from hp_tuning import NestedCVOptimizer
```

```python
files = ['two_class_raw_3s_no.csv', 'two_class_raw_3s_yo_0.5.csv',
        'two_class_raw_5s_no.csv', 'two_class_raw_5s_yo_0.5.csv']

base_path = 'improved_features_data_main/'
```

```python
all_results = {}
for file in files:
    data_path = os.path.join(base_path, file)
    features = pd.read_csv(data_path)
    features.drop(['center_time', 'start_time', 'end_time'], axis=1, inplace=True)
    details = file.split('_')
    exp_name = f"{details[3]}_{details[-1].replace('.csv', '')}"
    print(f"Analysing {exp_name}")
    
    # split data
    X = features.drop(columns=['label', 'experiment_id'])
    y = features['label']
    groups = features['experiment_id']
    
    # Initialize and run nested CV
    optimizer = NestedCVOptimizer(
        X=X, 
        y=y, 
        groups=groups,
        n_outer_folds=5,
        n_inner_folds=3,
        n_trials=100, 
        random_state=42
    )
    
    # Run nested cross-validation
    results = optimizer.run_nested_cv()
    all_results[exp_name] = results
```

```python
# pickle
# pickle the file
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/imb_files/hpt_results_100_trials.pkl', 'wb') as f:
    pickle.dump(all_results, f)
```
