```python
import pickle
import pandas as pd
```

```python
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/3_class_dt_cv_results_5_fold_stratified.pkl', 'rb') as f:
    dt_results = pickle.load(f)
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/3_class_rf_cv_results_5_fold_stratified.pkl', 'rb') as f:
    rf_results = pickle.load(f)
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/3_class_xgb_cv_results_5_fold_stratified.pkl', 'rb') as f:
    xgb_results = pickle.load(f)
```

```python
imb_tech = ['none', 'oversample_smote', 'undersample_tomek_links', 'combined_smote_tomek']
models = ['Decision Tree', 'Random Forest', 'XGBoost']
```

```python
# Combine all model results into one dictionary
all_results = {
    'Decision Tree': dt_results,
    'Random Forest': rf_results,
    'XGBoost': xgb_results
}
```

```python
def format_cv_results_separate_models(model_results_dict, models):
    """
    Convert cross-validation results from separate model dictionaries to tabular format.
    
    Parameters:
    model_results_dict: Dictionary where keys are model names and values are 
                    dictionaries with structure {dataset: {imbalance_technique: {metrics...}}}
    imb_tech: List of imbalance handling techniques
    models: List of model names
    
    Returns:
    pandas.DataFrame: Formatted results table
    """
    
    formatted_data = []
    
    # Iterate through each model
    for model in models:
        if model in model_results_dict:
            model_data = model_results_dict[model]
            
            
            # Iterate through each dataset for this model
            for dataset, dataset_results in model_data.items():
                
                # # Iterate through each imbalance technique
                # for technique in imb_tech:
                #     # Check if this technique exists in the results
                #     if technique in dataset_results:
                #         technique_results = dataset_results[technique]
                accuracy_avg = dataset_results.get('mean_accuracy', 0)
                accuracy_std = dataset_results.get('std_accuracy', 0)   
                macro_f1_avg = dataset_results.get('mean_macro_f1', 0)
                macro_f1_std = dataset_results.get('std_macro_f1', 0)
                macro_precision_avg = dataset_results.get('mean_macro_precision', 0)
                macro_precision_std = dataset_results.get('std_macro_precision', 0)
                macro_recall_avg = dataset_results.get('mean_macro_recall', 0)
                macro_recall_std = dataset_results.get('std_macro_recall', 0)
                weighted_f1_avg = dataset_results.get('mean_weighted_f1', 0)
                weighted_f1_std = dataset_results.get('std_weighted_f1', 0)
                weighted_precision_avg = dataset_results.get('mean_weighted_precision', 0)
                weighted_precision_std = dataset_results.get('std_weighted_precision', 0)
                weighted_recall_avg = dataset_results.get('mean_weighted_recall', 0)
                weighted_recall_std = dataset_results.get('std_weighted_recall', 0)
                
                precision_pre_void_avg = dataset_results.get('mean_precision_pre_void', 0)
                precision_pre_void_std = dataset_results.get('std_precision_pre_void', 0)
                recall_pre_void_avg = dataset_results.get('mean_recall_pre_void', 0)
                recall_pre_void_std = dataset_results.get('std_recall_pre_void', 0) 
                f1_pre_void_avg = dataset_results.get('mean_f1_pre_void', 0)
                f1_pre_void_std = dataset_results.get('std_f1_pre_void', 0)
                
                precision_void_avg = dataset_results.get('mean_precision_void', 0)
                precision_void_std = dataset_results.get('std_precision_void', 0)   
                recall_void_avg = dataset_results.get('mean_recall_void', 0)
                recall_void_std = dataset_results.get('std_recall_void', 0)
                f1_void_avg = dataset_results.get('mean_f1_void', 0)
                f1_void_std = dataset_results.get('std_f1_void', 0)
                
                precision_post_void_avg = dataset_results.get('mean_precision_post_void', 0)
                precision_post_void_std = dataset_results.get('std_precision_post_void', 0)
                recall_post_void_avg = dataset_results.get('mean_recall_post_void', 0)
                recall_post_void_std = dataset_results.get('std_recall_post_void', 0)
                f1_post_void_avg = dataset_results.get('mean_f1_post_void', 0)
                f1_post_void_std = dataset_results.get('std_f1_post_void', 0)
                
                
            
                # Add row to formatted data
                formatted_data.append({
                        'Dataset': dataset,
                        'Model': model,
                        'Accuracy (avg)': round(accuracy_avg, 5),
                        'Accuracy (std)': round(accuracy_std, 5),
                        'Accuracy (avg:std)': round(accuracy_avg/accuracy_std, 5),
                        'Macro F1 (avg)': round(macro_f1_avg, 5),
                        'Macro F1 (std)': round(macro_f1_std, 5),
                        'Macro F1 (avg:std)': round(macro_f1_avg/macro_f1_std, 5),
                        'Macro Precision (avg)': round(macro_precision_avg, 5),
                        'Macro Precision (std)': round(macro_precision_std, 5),
                        'Macro Precision (avg:std)': round(macro_precision_avg/macro_precision_std, 5),
                        'Macro Recall (avg)': round(macro_recall_avg, 5),
                        'Macro Recall (std)': round(macro_recall_std, 5),
                        'Macro Recall (avg:std)': round(macro_recall_avg/macro_recall_std, 5),
                        'Weighted F1 (avg)': round(weighted_f1_avg, 5),
                        'Weighted F1 (std)': round(weighted_f1_std, 5),
                        'Weighted F1 (avg:std)': round(weighted_f1_avg/weighted_f1_std, 5),
                        'Weighted Precision (avg)': round(weighted_precision_avg, 5),
                        'Weighted Precision (std)': round(weighted_precision_std, 5),
                        'Weighted Precision (avg:std)': round(weighted_precision_avg/weighted_precision_std, 5),
                        'Weighted Recall (avg)': round(weighted_recall_avg, 5),
                        'Weighted Recall (std)': round(weighted_recall_std, 5),
                        'Weighted Recall (avg:std)': round(weighted_recall_avg/weighted_recall_std, 5),
                        
                        'Precision Pre-Void (avg)': round(precision_pre_void_avg, 5),
                        'Precision Pre-Void (std)': round(precision_pre_void_std, 5),
                        'Precision Pre-Void (avg:std)': round(precision_pre_void_avg/precision_pre_void_std, 5),
                        'Recall Pre-Void (avg)': round(recall_pre_void_avg, 5),
                        'Recall Pre-Void (std)': round(recall_pre_void_std, 5),
                        'Recall Pre-Void (avg:std)': round(recall_pre_void_avg/recall_pre_void_std, 5),
                        'F1-Score Pre-Void (avg)': round(f1_pre_void_avg, 5),       
                        'F1-Score Pre-Void (std)': round(f1_pre_void_std, 5),
                        'F1-Score Pre-Void (avg:std)': round(f1_pre_void_avg/f1_pre_void_std, 5),   
                        
                        'Precision Void (avg)': round(precision_void_avg, 5),
                        'Precision Void (std)': round(precision_void_std, 5),
                        'Precision Void (avg:std)': round(precision_void_avg/precision_void_std, 5),
                        'Recall Void (avg)': round(recall_void_avg, 5),
                        'Recall Void (std)': round(recall_void_std, 5),
                        'Recall Void (avg/std)': round(recall_void_avg/recall_void_std, 5) ,
                        'F1-Score Void (avg)': round(f1_void_avg, 5),
                        'F1-Score Void (std)': round(f1_void_std, 5),
                        'F1-Score Void (avg:std)': round(f1_void_avg/f1_void_std, 5),
                        
                        
                        'Precision Post-Void (avg)': round(precision_post_void_avg, 5),
                        'Precision Post-Void (std)': round(precision_post_void_std, 5),
                        'Precision Post-Void (avg:std)': round(precision_post_void_avg/precision_post_void_std, 5),
                        'Recall Post-Void (avg)': round(recall_post_void_avg, 5),
                        'Recall Post-Void (std)': round(recall_post_void_std, 5),
                        'Recall Post-Void (avg:std)': round(recall_post_void_avg/recall_post_void_std, 5),  
                        'F1-Score Post-Void (avg)': round(f1_post_void_avg, 5),
                        'F1-Score Post-Void (std)': round(f1_post_void_std, 5),
                        'F1-Score Post-Void (avg:std)': round(f1_post_void_avg/f1_post_void_std, 5),
                        
                    })
                        
                        
    
    # Create DataFrame
    df = pd.DataFrame(formatted_data)
    
    # Sort by Dataset, Model, and Imbalance Technique for better organization
    # df = df.sort_values(['Dataset', 'Model', 'Imbalance Technique']).reset_index(drop=True)
    
    return df


print("Function ready to use for separate model dictionaries.")
```
