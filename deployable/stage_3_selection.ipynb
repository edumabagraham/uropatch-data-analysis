```python
import pickle
import pandas as pd
```

```python
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/dt_cv_results_5_fold_stratified.pkl', 'rb') as f:
    dt_results = pickle.load(f)
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/rf_cv_results_5_fold_stratified.pkl', 'rb') as f:
    rf_results = pickle.load(f)
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/training_results/xgb_cv_results_5_fold_stratified.pkl', 'rb') as f:
    xgb_results = pickle.load(f)
```

```python
imb_tech = ['none', 'oversample_smote', 'undersample_tomek_links', 'combined_smote_tomek']
models = ['Decision Tree', 'Random Forest', 'XGBoost']
```

```python
# Combine all model results into one dictionary
all_results = {
    'Decision Tree': dt_results,
    'Random Forest': rf_results,
    'XGBoost': xgb_results
}
```

```python
def format_cv_results_separate_models(model_results_dict, imb_tech, models):
    """
    Convert cross-validation results from separate model dictionaries to tabular format.
    
    Parameters:
    model_results_dict: Dictionary where keys are model names and values are 
                    dictionaries with structure {dataset: {imbalance_technique: {metrics...}}}
    imb_tech: List of imbalance handling techniques
    models: List of model names
    
    Returns:
    pandas.DataFrame: Formatted results table
    """
    
    formatted_data = []
    
    # Iterate through each model
    for model in models:
        if model in model_results_dict:
            model_data = model_results_dict[model]
            
            
            # Iterate through each dataset for this model
            for dataset, dataset_results in model_data.items():
                
                # Iterate through each imbalance technique
                for technique in imb_tech:
                    # Check if this technique exists in the results
                    if technique in dataset_results:
                        technique_results = dataset_results[technique]
                        
                        recall_void_avg = technique_results.get('mean_recall_minority', 0)
                        recall_void_std = technique_results.get('std_recall_minority', 0)
                        accuracy_avg = technique_results.get('mean_accuracy', 0)
                        accuracy_std = technique_results.get('std_accuracy', 0)
                        f1_void_avg = technique_results.get('mean_f1_minority', 0)
                        f1_void_std = technique_results.get('std_f1_minority', 0)
                        macro_f1_avg = technique_results.get('macro_f1', 0)
                        macro_f1_std = technique_results.get('std_macro_f1', 0)
                        weighted_f1_avg = technique_results.get('weighted_f1', 0)
                        weighted_f1_std = technique_results.get('std_weighted_f1', 0)
                        
                        if technique == "none":
                            technique = "Baseline (none)"
                        elif technique == "oversample_smote":
                            technique = "SMOTE"
                        elif technique == "undersample_tomek_links":
                            technique = "TomekLinks"
                        elif technique == "combined_smote_tomek":
                            technique = "SMOTETomek"
                        
                        # Add row to formatted data
                        formatted_data.append({
                            'Dataset': dataset,
                            'Model': model,
                            'Imbalance Technique': technique,
                            'F1-Score (avg)': round(f1_void_avg, 5),
                            'F1-Score (std)': round(f1_void_std, 5),
                            'F1-Score (avg:std)': round(f1_void_avg/f1_void_std, 5),
                            'Recall (avg)': round(recall_void_avg, 5),
                            'Recall (std)': round(recall_void_std, 5),
                            'Recall (avg:std)': round(recall_void_avg/recall_void_std, 5),
                            'Accuracy (avg)': round(accuracy_avg, 5),
                            'Accuracy (std)': round(accuracy_std, 5),
                            'Accuracy (avg:std)': round(accuracy_avg/accuracy_std, 5),
                            'Macro F1 (avg)': round(macro_f1_avg, 5),
                            'Macro F1 (std)': round(macro_f1_std, 5),
                            'Macro F1 (avg:std)': round(macro_f1_avg/macro_f1_std, 5),
                            'Weighted F1 (avg)': round(weighted_f1_avg, 5),
                            'Weighted F1 (std)': round(weighted_f1_std, 5),
                            'Weighted F1 (avg:std)': round(weighted_f1_avg/weighted_f1_std, 5)
                            
                        })
                        
                        
    
    # Create DataFrame
    df = pd.DataFrame(formatted_data)
    
    # Sort by Dataset, Model, and Imbalance Technique for better organization
    # df = df.sort_values(['Dataset', 'Model', 'Imbalance Technique']).reset_index(drop=True)
    
    return df


print("Function ready to use for separate model dictionaries.")
```

```python
df = format_cv_results_separate_models(all_results, imb_tech, models)
df
```

```python
df_sorted_f1_score = df.sort_values(by='F1-Score (avg)', ascending=False).reset_index(drop=True)
df_sorted_f1_score
```
