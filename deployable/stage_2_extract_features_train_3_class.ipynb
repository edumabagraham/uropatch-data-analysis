```python
# Save the dictionay
with open('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/train_data_with_3_class_labels.pkl', 'rb') as f:
    train_data_3_class_labelled = pickle.load(f)
```

## Data 1
classes: void, non-void

fs: original

sliding window: 1s

overlap: 0

filename: two_class_1s_no.csv

```python
#  do not upsample
all_features = []
for i, exp_id in enumerate(train_data_3_class_labelled.keys()):    
    # Extract features
    
    df = train_data_3_class_labelled[exp_id]
    fs = 1 / df['time'].diff().median()
    analyzer = GenerateFeatures(fs, window_duration=1.0, overlap=0.0)
    features, _ = analyzer.analyze_multi_axis_imu(train_data_3_class_labelled[exp_id])

    table = analyzer.create_summary_table()
    table['experiment_id'] = exp_id 
    all_features.append(table)
    

final_features = pd.concat(all_features, ignore_index=True)
```

```python
final_features.to_csv('/home/edumaba/Public/MPhil_Thesis/Code/uropatch-data-analysis/deployable/data/extracted_features_data/3_class/train_set/two_class_1s_no.csv', index=False)
final_features
```
